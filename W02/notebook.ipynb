````python
print(2 + 2)
print("2" + "2")
````

One printed `4`. The other printed `22` &mdash; not `4`, and not `"4"`, but the characters `2` and `2` stuck end to end.

The `+` symbol did not change. What changed is **what it was given**. On the first line it saw two integers and *added* them. On the second it saw two strings and *joined* them. Same operator, two behaviors, chosen entirely by the **type** of the values on either side.

Before we can talk about that properly, we need a way to ask Python what type a value actually is.

## Asking Python What Something Is: `type()`

You do not have to guess. Python will tell you the type of any value if you hand it to the built-in `type()`.

````python
print(type(42))
print(type(3.14))
print(type("COMP 170"))
print(type(2 + 2 == 4))
````

The output:

```
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

Four types we will use constantly:

<table class="cmp-table">
<tr><th>Type</th><th>Python name</th><th>What it holds</th><th>Examples</th></tr>
<tr><td>Integer</td><td><code>int</code></td><td>whole numbers, positive or negative</td><td><code>42</code>, <code>0</code>, <code>-7</code></td></tr>
<tr><td>Floating-point</td><td><code>float</code></td><td>numbers with a decimal point</td><td><code>3.14</code>, <code>-0.5</code>, <code>100.0</code></td></tr>
<tr><td>String</td><td><code>str</code></td><td>text, in quotes</td><td><code>"hello"</code>, <code>'COMP 170'</code></td></tr>
<tr><td>Boolean</td><td><code>bool</code></td><td>one of exactly two values: <code>True</code> or <code>False</code></td><td><code>3 &gt; 5</code> &rarr; <code>False</code></td></tr>
</table>

That last one is new. A **comparison** like `2 + 2 == 4` is a question with a yes/no answer, and Python answers it with a `bool`. (`==`, two equals signs, asks *"are these equal?"* &mdash; it is not the single `=` that assigns a variable. Much more on this in Week 6.) For now just notice that `True` and `False` are values with a type, exactly like `42` and `"hello"`.

````python
course = "COMP 170"
credits = 3
enrolled = credits >= 1

print(type(course))
print(type(credits))
print(type(enrolled))
````

<div class="keyterm">

Reach for it whenever a value behaves in a way you did not expect. The <code>&lt;class '...'&gt;</code> wrapper just means "type"; the useful part is the name inside the quotes.

## The Same Operator, Chosen by Type: Operator Overloading

Now we can name what we saw at the start. `2 + 2` and `"2" + "2"` behave differently because `+` is **overloaded**: it carries more than one meaning, and Python picks which one to run based on the types of its operands.

<table class="cmp-table">
<tr><th>Expression</th><th>Types</th><th>What <code>+</code> does</th><th>Result</th></tr>
<tr><td><code>2 + 2</code></td><td><code>int</code>, <code>int</code></td><td>arithmetic addition</td><td><code>4</code></td></tr>
<tr><td><code>"2" + "2"</code></td><td><code>str</code>, <code>str</code></td><td>concatenation (joining)</td><td><code>"22"</code></td></tr>
<tr><td><code>"Hello, " + "Ada"</code></td><td><code>str</code>, <code>str</code></td><td>concatenation</td><td><code>"Hello, Ada"</code></td></tr>
<tr><td><code>2 + "2"</code></td><td><code>int</code>, <code>str</code></td><td>no agreed meaning &mdash; Python refuses</td><td><code>TypeError</code></td></tr>
</table>

````python
print("Hello, " + "Ada")
print("COMP" + " " + "170")

# Uncomment the next line and run again. Read the error message all the way through.
# print(2 + "2")
````

<div class="keyterm">
<strong>Definition &mdash; operator overloading:</strong> when a single operator symbol (like <code>+</code>) performs different operations depending on the types of the values it is applied to. <code>+</code> on numbers adds; <code>+</code> on strings concatenates. When the types share no meaning for that operator &mdash; <code>2 + "2"</code> &mdash; Python raises a <code>TypeError</code> rather than guessing. It will not assume you meant <code>"2"</code> to be the number 2; if you want that, you have to say so.
</div>

## Arithmetic: The Full Set of Operators

We used `+` and `**` last week without a full tour. Here is the whole toolkit. Predict all seven results before running:

````python
print(7 + 2)    # addition
print(7 - 2)    # subtraction
print(7 * 2)    # multiplication
print(7 / 2)    # division       -> always a float
print(7 // 2)   # floor division -> drops the fractional part
print(7 % 2)    # modulo         -> the remainder
print(7 ** 2)   # exponentiation -> 7 to the power 2
````

The one that catches everyone is `7 / 2` giving `3.5` when `7 // 2` gives `3`. In Python 3, a single `/` **always produces a `float`**, even when it divides evenly. Here is last week's cliffhanger, finally:

````python
print(10 / 2)   # 5.0  -- still a float, even though it is exact
print(10 / 3)   # 3.3333333333333335  -- does NOT come out exact
print(10 // 3)  # 3    -- floor division keeps only the whole part
print(10 % 3)   # 1    -- ...and modulo keeps what floor division threw away
````

So `10 / 3` does not come out exact: it comes out as a `float` printed to many digits. That is the trade a `float` makes &mdash; it can represent `3.5` and `0.001`, and it pays for that range by giving up exactness. Use `//` when you want the whole-number part and `%` when you want the remainder.

<div class="keyterm">
<strong>Definitions:</strong><br>
<code>/</code> &mdash; true division, always returns a <code>float</code> (<code>7 / 2</code> &rarr; <code>3.5</code>).<br>
<code>//</code> &mdash; floor division, discards the fractional part (<code>7 // 2</code> &rarr; <code>3</code>).<br>
<code>%</code> &mdash; modulo, the remainder after floor division (<code>7 % 2</code> &rarr; <code>1</code>).<br>
<code>**</code> &mdash; exponentiation (<code>7 ** 2</code> &rarr; <code>49</code>).
</div>

## When Operators Share a Line: Precedence

Predict this one, then run it:

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
````

`2 + 3 * 4` is `14`, not `20`. Python did the multiplication first. It follows the same **order of operations** you learned in arithmetic class:

1. Parentheses `( )`
2. Exponentiation `**`
3. `*`, `/`, `//`, `%`  (left to right)
4. `+`, `-`  (left to right)

So in a formula like `principal * (1 + rate) ** years`, the parentheses do real work: they force `1 + rate` to happen *before* the exponentiation. Without them you would be computing `principal * 1 + (rate ** years)` &mdash; a different, wrong number.

<div class="keyterm">
<strong>Definition &mdash; operator precedence:</strong> the fixed order in which Python evaluates operators when a line contains more than one. Higher-precedence operators run first; equal-precedence operators go left to right. When in doubt, add parentheses: they cost nothing and they make your intent visible to the next reader.
</div>

## A Quick Distinction: Expressions and Statements

Two words we will use all term. Some lines of code **have a value** &mdash; `7 // 2`, `2 + 3 * 4`, `type(42)`, `"a" + "b"`. Those are **expressions**: you can always ask "what does this evaluate to?" and put the answer wherever a value is expected (on the right of `=`, inside `print(...)`, inside another expression).

Other lines just **do something**. `total = 14` is a **statement**: it changes what the name `total` points to, but it has no value of its own.

We will lean on this a lot next week, when a call like `future_value(1000, 0.045, 10)` turns out to be an expression too. For now: *expression = has a value; statement = performs an action.*

Recap of Lecture 1: every value has a type (`int`, `float`, `str`, `bool`); `type()` reports it; `+` is overloaded (add vs concatenate); `/` always gives a `float`, `//` and `%` give the whole part and the remainder; precedence decides what happens first on a shared line.

## Getting Input From a Person: `input()`

So far our programs always compute the same thing. `input()` lets a program pause and ask. Here is one that should double whatever number you give it. Run it and type `25`.

````python
age = input("Enter a number: ")
print("Doubled:", age * 2)
````

If you typed `25`, it printed `2525`, not `50`.

You can explain that from Lecture 1: `age * 2` repeated the string `"25"`. Because **`input()` always returns a `str`** &mdash; whatever the person typed, handed back as text, even when it looks like a number.

To do arithmetic on it, **convert** it first:

````python
age = input("Enter a number: ")
age = int(age)                 # "25" (str)  ->  25 (int)
print("Doubled:", age * 2)
````

Once you trust the pattern, the conversion and the input go on one line:

````python
age = int(input("Enter a number: "))
print("Doubled:", age * 2)
````

`int()`, `float()`, and `str()` are **type conversion** functions: hand them a value, get back the same information as a different type.

<table class="cmp-table">
<tr><th>Call</th><th>Result</th><th>Note</th></tr>
<tr><td><code>int("25")</code></td><td><code>25</code></td><td>text that looks like a whole number &rarr; <code>int</code></td></tr>
<tr><td><code>float("3.14")</code></td><td><code>3.14</code></td><td>text with a decimal point &rarr; <code>float</code></td></tr>
<tr><td><code>int(3.9)</code></td><td><code>3</code></td><td><strong>truncates</strong>, does not round</td></tr>
<tr><td><code>str(25)</code></td><td><code>"25"</code></td><td>number &rarr; text, e.g. to concatenate it</td></tr>
<tr><td><code>int("hello")</code></td><td><code>ValueError</code></td><td>no sensible conversion &mdash; Python refuses</td></tr>
<tr><td><code>int("3.9")</code></td><td><code>ValueError</code></td><td><code>int()</code> will not parse a decimal point from text; use <code>int(float("3.9"))</code></td></tr>
</table>

<div class="gotcha">
<strong>Two traps:</strong> (1) <code>int(3.9)</code> is <code>3</code>, not <code>4</code> &mdash; converting to <code>int</code> chops the fractional part off, it never rounds. (2) <code>int("hello")</code> does not return <code>0</code> or <code>None</code> &mdash; it stops the program with a <code>ValueError</code>. That is Python reporting an impossible conversion, not papering over it.
</div>

<div class="keyterm">
<strong>Definition &mdash; type conversion:</strong> using <code>int()</code>, <code>float()</code>, or <code>str()</code> to build a new value of a target type from an existing value. Needed right after <code>input()</code>, whose result is always a <code>str</code>.
</div>

````python
print(int(3.9))        # 3  -- truncated, not rounded

# Uncomment each in turn, run, and read the error:
# print(int("hello"))
# print(int("3.9"))
````

## The Error Python Won't Catch For You

Last week every error we hit stopped the program: a missing parenthesis, a name used before it was defined. Python noticed and complained. Here is a program to average two test scores. Run it with `80` and `90`.

````python
a = int(input("First score: "))
b = int(input("Second score: "))
average = a + b / 2
print("Average:", average)
````

With `80` and `90` it prints `Average: 125.0`.

Nobody's average of 80 and 90 is 125. But Python said nothing &mdash; no red text, no traceback, no crash. The program is **syntactically perfect** and it **ran to completion**. It just computed the wrong thing, because by precedence `a + b / 2` means `a + (b / 2)`, not `(a + b) / 2`.

This is a **logic error**: the code does exactly what you wrote, and what you wrote is not what you meant. It is a genuinely different category from last week's errors.

<table class="cmp-table">
<tr><th>Kind of error</th><th>When it is caught</th><th>What you see</th><th>Example</th></tr>
<tr><td><strong>Syntax error</strong></td><td><em>Before</em> the program runs &mdash; Python cannot even parse it</td><td>The program never starts; Python points at the line</td><td><code>print("hi"</code> &nbsp;(missing <code>)</code>)</td></tr>
<tr><td><strong>Runtime error</strong></td><td><em>During</em> the run, at the offending line</td><td>The program starts, then stops partway with a traceback</td><td><code>int("hello")</code>, &nbsp;<code>2 + "2"</code></td></tr>
<tr><td><strong>Logic error</strong></td><td><em>Never</em>, by Python &mdash; only by you</td><td>The program runs cleanly and returns a confident wrong answer</td><td><code>a + b / 2</code> for an average</td></tr>
</table>

Python finds the first two for you. The third is yours to catch: know what answer to expect, test with inputs whose result you can check by hand (the average of 80 and 90 is 85), and read your own code the way that literal cook from Week 1 would.

````python
a = int(input("First score: "))
b = int(input("Second score: "))
average = (a + b) / 2      # parentheses force the addition first
print("Average:", average)
````

## Naming Things Well

Last week we covered the *rules* for identifiers: start with a letter or underscore; letters, digits, underscores only; no keywords. Those say what Python will *accept*. They do not say what a human reading your code at 2 am will *thank you for*.

<table class="cmp-table">
<tr><th>Accepted by Python</th><th>Also accepted by Python</th></tr>
<tr>
<td><pre><code>x = int(input("Score: "))
y = int(input("Score: "))
z = (x + y) / 2
print(z)</code></pre></td>
<td><pre><code>first_score  = int(input("Score: "))
second_score = int(input("Score: "))
average = (first_score + second_score) / 2
print(average)</code></pre></td>
</tr>
</table>

Same logic, same output. The right-hand version explains itself. The conventions Python programmers follow (written down as **PEP 8**):

- **`snake_case`** &mdash; lowercase words joined by underscores: `first_score`, `total_with_tax`. Not `firstScore`.
- **Descriptive over short** &mdash; `principal`, not `p`. A throwaway loop counter `i` is fine; a value that matters deserves a real name.
- **`UPPER_CASE` for constants** &mdash; values set once and never changed: `TAX_RATE = 0.0825`.
- **Don't shadow built-ins** &mdash; naming a variable `type`, `str`, `sum`, or `input` replaces that tool for the rest of the program.

<div class="keyterm">
<strong>Definition &mdash; naming convention:</strong> a shared, non-mandatory agreement about how to format identifiers so code from different people reads consistently. Breaking it causes no error &mdash; only friction.
</div>
