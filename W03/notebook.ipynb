## Procedural decomposition

> **Make cookies** → **Prepare dough** + **Bake one tray** + **Bake one tray**

| 🔴 1 · Decompose | → | 🟢 2 · Name | → | 🟡 3 · Reuse |
|:---:|:---:|:---:|:---:|:---:|
| Divide a large process into smaller, meaningful jobs. | | Give each job a name that says what it does. | | Call the same job whenever it is needed. |

## Functions: named, reusable pieces of code

```python
def bake_tray():
    print("Preheat the oven to 350 degrees.")
    print("Place one tray in the oven.")
    print("Set the timer for 12 minutes.")
    print("Remove the tray.")
```

> **Historical note.** Early programmers used *subroutines* to name and reuse a sequence of instructions. Today, related ideas may be called **functions** or **methods**. Each packages a job so a program can call it by name.

````python
def prepare_dough():
    print("Mix the ingredients.")
    print("Shape the cookies.")


def bake_tray():
    print("Preheat the oven to 350 degrees.")
    print("Place one tray in the oven.")
    print("Set the timer for 12 minutes.")
    print("Remove the tray.")


prepare_dough()
bake_tray()
bake_tray()
````

> Procedural decomposition separates a large process into smaller jobs. A function stores the instructions for one of those jobs under a meaningful name. In the recipe, the baking instructions are declared once but called twice. This removes duplicated code, makes the overall process easier to read, and gives us one place to update the baking steps.

````python
# DECLARATION: teach Python a new named job.
def greet():                  # FUNCTION SIGNATURE
    message = "Hello"         # BODY + LOCAL VARIABLE
    print(message)            # BODY


print("Before")
greet()                       # CALL: run the named job.
# When greet() finishes, execution returns here to its caller.
print("After")
````

|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| Create the named job | | Request the job | | Execute its statements | | Continue after the call |

## Anatomy of a declaration

```python
def greet():                 # function signature
    message = "Hello!"       # function body
    print(message)           # function body
```

| 🔵 Signature | 🟢 Body | 🟡 Local variable |
|:---:|:---:|:---:|
| `def`, name, parentheses, colon | Indented statements that run when called | Created inside; usable only inside that function |

````python
def greet():
    message = "Hello!"
    print(message)


print("Before the call")
greet()
print("After the call")

# print(message)  # NameError: message is local to greet()
````

## Functions can call functions

```python
def make_breakfast():       # caller
    toast_bread()            # calls toast_bread
    pour_juice()             # calls pour_juice
```

| 🔵 **Caller** | → | 🟢 **Called function** | → | 🟡 **Back to caller** |
|:---:|:---:|:---:|:---:|:---:|
| `make_breakfast()` | | `toast_bread()` | | Continue inside `make_breakfast()` |

- A function may be both a **caller** and a **called function**.
- Each call temporarily transfers control to another function.

````python
def toast_bread():
    print("Toast the bread.")


def pour_juice():
    print("Pour the juice.")


def make_breakfast():
    print("Breakfast begins.")
    toast_bread()
    pour_juice()
    print("Breakfast is ready.")


make_breakfast()
````

## You have been calling functions already

| 🔵 `print(...)` | 🟢 `input(...)` | 🟡 `int(...)` |
|:---:|:---:|:---:|
| Display information | Read text | Convert to an integer |

Someone else declared these functions. We only need to know their names, what they expect, and what they produce.

> A function declaration begins with a signature and is followed by an indented body. Declaring a function makes the instructions available; calling it executes them. When one function calls another, the first is the caller and execution returns to it when the called function finishes. Parameters and variables created inside a function are local to that call and cannot normally be used outside it. Built-ins such as `print`, `input`, and `int` are functions already declared for us.

# Case study: comparing savings plans

*How should a program organize values, calculations, and output?*

| 🔵 Starter | 🟢 Growth | 🟡 Long-Term |
|:---:|:---:|:---:|
| $1,000 · 3% · 5 years | $1,500 · 4% · 8 years | $2,000 · 5% · 10 years |

````python
# Plan 1
plan_name = "Starter"
principal = 1000
rate = 0.03
years = 5
future_value = principal * (1 + rate) ** years
print("PLAN:", plan_name)
print("Deposit:", principal, "Rate:", rate, "Years:", years)
print("Future value:", round(future_value, 2))

# Plan 2
plan_name = "Growth"
principal = 1500
rate = 0.04
years = 8
future_value = principal * (1 + rate) ** years
print("PLAN:", plan_name)
print("Deposit:", principal, "Rate:", rate, "Years:", years)
print("Future value:", round(future_value, 2))

# Plan 3
plan_name = "Long-Term"
principal = 2000
rate = 0.05
years = 10
future_value = principal * (1 + rate) ** years
print("PLAN:", plan_name)
print("Deposit:", principal, "Rate:", rate, "Years:", years)
print("Future value:", round(future_value, 2))
````

## Correct result, weak structure

| 🔵 Hard to scan | 🔴 Hard to change | 🟡 Hidden jobs |
|:---:|:---:|:---:|
| Three plans form one long sequence. | The same kind of work appears in several places. | “Process one plan” has no name. |

**First step:** give each plan a named piece of code.

````python
def show_starter_plan():
    principal = 1000
    rate = 0.03
    years = 5
    value = principal * (1 + rate) ** years
    print("PLAN: Starter")
    print("Deposit:", principal, "Rate:", rate, "Years:", years)
    print("Future value:", round(value, 2))


def show_growth_plan():
    principal = 1500
    rate = 0.04
    years = 8
    value = principal * (1 + rate) ** years
    print("PLAN: Growth")
    print("Deposit:", principal, "Rate:", rate, "Years:", years)
    print("Future value:", round(value, 2))


def show_long_term_plan():
    principal = 2000
    rate = 0.05
    years = 10
    value = principal * (1 + rate) ** years
    print("PLAN: Long-Term")
    print("Deposit:", principal, "Rate:", rate, "Years:", years)
    print("Future value:", round(value, 2))


show_starter_plan()
show_growth_plan()
show_long_term_plan()
````

## What differs among the three functions?

> The instructions are the same. **Only the values change.**

| Plan name | + | Principal | + | Rate | + | Years | → | **One general function** |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|

What if the caller could pass those values into the function?

````python
def show_plan(plan_name, principal, rate, years):
    value = principal * (1 + rate) ** years
    print("PLAN:", plan_name)
    print("Deposit:", principal, "Rate:", rate, "Years:", years)
    print("Future value:", round(value, 2))


show_plan("Starter", 1000, 0.03, 5)
show_plan("Growth", 1500, 0.04, 8)
show_plan("Long-Term", 2000, 0.05, 10)
````

## Parameters and arguments

```python
def function_name(parameter_1, parameter_2):
    # use the parameters

function_name(argument_1, argument_2)
```

| Parameters | Arguments |
|---|---|
| Names in the function declaration | Actual values supplied by a call |
| `plan_name, principal, rate, years` | `"Starter", 1000, 0.03, 5` |

## Each call gets its own local values

| Call | `plan_name` | `principal` | `rate` | `years` |
|---|---:|---:|---:|---:|
| Call 1 | `"Starter"` | `1000` | `0.03` | `5` |
| Call 2 | `"Growth"` | `1500` | `0.04` | `8` |
| Call 3 | `"Long-Term"` | `2000` | `0.05` | `10` |

The same body runs three times with three different sets of values.

> Parameters let one function describe a general job. They are names listed in the declaration and behave like local variables during each call. Arguments are the actual values supplied by the caller. In positional calls, Python matches arguments to parameters from left to right. Replacing three nearly identical plan functions with one parameterized function keeps one copy of the instructions while still supporting different plans.

## One function—two concerns

```python
def show_plan(plan_name, principal, rate, years):
    value = principal * (1 + rate) ** years   # calculate
    print(...)                                # display
```

| 🟢 Concern 1 | 🟡 Concern 2 |
|:---:|:---:|
| Calculate the future value. | Format and display a plan. |

> **Design guideline:** try to give each function one clear concern.

## The missing connection

If calculation and display are separate...

| `calculate_future_value(...)` | → | **?** | → | `display_plan(...)` |
|:---:|:---:|:---:|:---:|:---:|

How can the calculation function send its result back to its caller?

> **`return` gives a function an output.**

````python
def calculate_future_value(principal, rate, years):
    value = principal * (1 + rate) ** years
    return value


future_value = calculate_future_value(1000, 0.03, 5)
print(future_value)
````

## Follow the returned value

| 🔵 ① Call | → | 🟢 ② Calculate | → | 🟡 ③ Return | → | 🟣 ④ Receive |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| `calculate_future_value(1000, .03, 5)` | | `1159.27...` | | `return value` | | `future_value = ...` |

- `return` ends that function call.
- The returned value replaces the call expression.


## Three ways to use a returned value

```python
# 1. Store it
total = calculate_future_value(1000, 0.03, 5)

# 2. Display it directly
print(calculate_future_value(1000, 0.03, 5))

# 3. Use it in a larger expression
doubled = 2 * calculate_future_value(1000, 0.03, 5)
```

The function call is an expression: it produces a value.


## `print` and `return` have different jobs

| `print(value)` | `return value` |
|---|---|
| Displays information for a person | Sends a value to the caller |
| The program cannot reuse the displayed text as the result | The caller can store, print, or combine the result |
| Does not end the function | Ends the current function call |

## Separate input, calculation, and output

| 🔵 **Input** | → | 🟢 **Logic** | → | 🟡 **Output** |
|:---:|:---:|:---:|:---:|:---:|
| `read_plan()` | | `calculate_future_value()` | | `display_plan()` |

| Function | One concern |
|---|---|
| `read_plan` | Prompt for and return the plan values |
| `calculate_future_value` | Calculate and return one number |
| `display_plan` | Receive and display the values |

````python
def read_plan():
    name = input("Plan name: ")
    principal = float(input("Starting amount: "))
    rate = float(input("Annual rate as a decimal: "))
    years = int(input("Number of years: "))
    return name, principal, rate, years


def calculate_future_value(principal, rate, years):
    return principal * (1 + rate) ** years


def display_plan(name, principal, rate, years, future_value):
    print("PLAN:", name)
    print("Deposit:", principal, "Rate:", rate, "Years:", years)
    print("Future value:", round(future_value, 2))
````

````python
def main():
    name, principal, rate, years = read_plan()
    future_value = calculate_future_value(principal, rate, years)
    display_plan(name, principal, rate, years, future_value)


main()
````

## What does `main()` mean in Python?

| 🟢 Coordinator | 🔵 Conventional name | 🟡 Front door |
|:---:|:---:|:---:|
| Shows the major steps and how functions cooperate | Python does not automatically treat `main` as special | We call it to start the program's work |

```python
def main():
    ...

main()
```

Other languages may give `main` a formally defined role; in Python, it is a useful convention.

> A well-focused function usually has one concern. Separating input, calculation, and output makes each part easier to read, test, and reuse. The `return` statement sends a value to the caller and ends the current call. Because the call expression becomes that returned value, the caller can assign it, print it, or use it inside another expression. A function with no explicit `return` produces Python's special value `None`. The conventional `main()` function coordinates the program's major jobs; it becomes the starting point only because we call it.

# Plan first, then translate

### Pseudocode → docstring → type-annotated Python

> **When code becomes difficult to follow, pause before adding more Python. Write the plan in plain language.**


## Pseudocode: a plan for the program

```text
PROGRAM: Compare one savings plan

Read the plan name, principal, rate, and years.
Calculate the future value.
Display the plan details and future value.
```

- Focuses on the logic, not Python punctuation.
- Reveals the major jobs and their order.
- Can be reviewed before implementation begins.


## A contract for each function

| Purpose | Preconditions | Postconditions |
|---|---|---|
| What job does it perform? | What inputs does it expect? What constraints must they satisfy? | What result or effect does it guarantee? |

```text
FUNCTION calculate_future_value(principal, rate, years)
PURPOSE: calculate value after annual compounding
PRE: principal >= 0, rate >= 0, years >= 0
POST: return the calculated future value
```

## Put the plain-English contract beside the code

```python
def calculate_future_value(principal, rate, years):
    """Calculate value after annual compounding.

    Preconditions: principal, rate, and years are nonnegative.
    Postcondition: return the calculated future value.
    """
    return principal * (1 + rate) ** years
```

> A **docstring** is documentation stored inside a function, immediately below its signature.

## Type annotations clarify the signature

```python
def calculate_future_value(
    principal: float,
    rate: float,
    years: int
) -> float:
```

| 🔵 Parameter annotations | 🟢 Return annotation | 🟡 Important |
|:---:|:---:|:---:|
| `principal: float`<br>Expected input type | `-> float`<br>Expected output type | Annotations document expectations; Python does not normally enforce them by itself. |

````python
def calculate_future_value(
    principal: float,
    rate: float,
    years: int
) -> float:
    """Calculate value after annual compounding.

    Preconditions:
        principal, rate, and years are nonnegative.
    Postcondition:
        Return the calculated future value.
    """
    return principal * (1 + rate) ** years


print(calculate_future_value(1000.0, 0.03, 5))
help(calculate_future_value)
````

## One idea, three representations

| Pseudocode | Docstring | Type annotations |
|---|---|---|
| Design the logic before coding | Keep the purpose and contract with the code | Make expected input/output types visible in the signature |

| 🔵 Think | → | 🟢 Document | → | 🟡 Implement | → | 🟣 Revise |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|

> Pseudocode is a language-independent plan: it states the program's jobs and their order before syntax becomes a distraction. At function level, the plan should identify the purpose, inputs and their constraints (preconditions), and the promised result or effect (postconditions). A docstring preserves that contract inside the implementation, where tools such as `help()` can display it. Type annotations complement the docstring by recording expected parameter and return types directly in the signature; they improve communication but do not automatically reject values of other types.

## When one value is usually the same

Most customers buy one copy. Can the function use `1` when the caller does not provide a quantity?

```python
def order_cost(price, quantity=1):
```

| 🔵 Argument omitted | 🟢 Argument supplied |
|:---:|:---:|
| `order_cost(18.0)`<br>`quantity` receives `1` | `order_cost(18.0, 4)`<br>`quantity` receives `4` |

A parameter with a **default value** becomes optional for the caller. Required parameters must come first.

````python
def order_cost(price: float, quantity: int = 1) -> float:
    """Return the total cost of a book order."""
    return price * quantity


print(order_cost(18.0))
print(order_cost(18.0, 4))
````

## Exercise 3 — Solution

One function can send more than one related value back to its caller.

````python
def temperature_range(
    morning: float,
    afternoon: float,
    evening: float
) -> tuple:
    """Return the lowest and highest temperatures."""
    lowest = min(morning, afternoon, evening)
    highest = max(morning, afternoon, evening)
    return lowest, highest


low, high = temperature_range(58.0, 74.0, 65.0)
print("Low:", low)
print("High:", high)
````

## Returning multiple values

```python
return lowest, highest

low, high = temperature_range(58, 74, 65)
```

| 🔵 Return two values | → | 🟢 Receive two values | → | 🟡 Use each separately |
|:---:|:---:|:---:|:---:|:---:|

Python bundles the returned values together. The assignment unpacks them into matching variables.

## Exercise 4 — One possible decomposition

| Function | Purpose | Inputs | Output |
|---|---|---|---|
| `read_order` | Read the order | None | Name and two page counts |
| `calculate_cost` | Calculate the total | Two page counts | Total cost |
| `display_receipt` | Display the result | Name, counts, total | None |
| `main` | Coordinate the program | None | None |

````python
def read_order() -> tuple:
    """Read and return one student's printing order."""
    name = input("Student name: ")
    black_white_pages = int(input("Black-and-white pages: "))
    color_pages = int(input("Color pages: "))
    return name, black_white_pages, color_pages


def calculate_cost(black_white_pages: int, color_pages: int) -> float:
    """Return the total printing cost."""
    return black_white_pages * 0.10 + color_pages * 0.50


def display_receipt(
    name: str,
    black_white_pages: int,
    color_pages: int,
    total: float
) -> None:
    """Display one student's printing receipt."""
    print("Student:", name)
    print("Black-and-white pages:", black_white_pages)
    print("Color pages:", color_pages)
    print("Total cost: $", round(total, 2))
````

````python
def main() -> None:
    """Coordinate the printing-kiosk program."""
    name, black_white_pages, color_pages = read_order()
    total = calculate_cost(black_white_pages, color_pages)
    display_receipt(name, black_white_pages, color_pages, total)


main()
````

> The exercises extend the core model in two directions. A default parameter supplies a value when the caller omits the corresponding argument. A function can also return several related results, which the caller can unpack into separate variables. The final exercise emphasizes decomposition: identify the program's distinct concerns, define a contract for each function, and let `main()` coordinate their calls.
