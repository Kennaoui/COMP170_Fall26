### Step 3: Safely open and exit Vim

Let's try a tiny, guided exercise together. In your terminal:

1. Type `vim week00_test.txt` and press Enter.
2. Vim opens in **Normal mode**.
3. Press `i` to switch to **Insert mode**.
4. Type: `COMP 170 is ready.`
5. Press `Escape` to return to Normal mode.
6. Type `:wq`
7. Press Enter — this saves the file and exits.

A couple of quick notes:

- `i` begins inserting text.
- `Escape` returns you to Normal mode.
- `:wq` saves ("writes") your file and quits.
- Want to exit *without* saving? Press `Escape`, type `:q!`, and press Enter.
- If Vim says "No write since last change," use `:wq` to save, or `:q!` to discard your changes.

> **You do not need to memorize any of this yet.** We'll practice it together in class.

We won't get into Vim configuration files, plugins, mouse settings, other modes, or advanced navigation here — that's all for another day.
