## What Pseudocode Is

<div class="keyterm">
<strong>pseudocode:</strong> an English-like description of an algorithm &mdash; not a real programming language, but structured enough that it converts easily into one.
</div>

Writing pseudocode means thinking a problem through *before* touching Python, so the hard thinking happens on paper, not while you're also fighting syntax errors.

- **Pseudocode comes first.** Writing the code, then describing it afterward, is a different (and much less useful) exercise &mdash; and it's graded as if no pseudocode was submitted at all.
- **It's not the same as comments added to finished code.** It's a plan that exists *before* the program does. Once it's solid, a good habit is leaving it in as comments above each function &mdash; free documentation, but only after it's done its job as a plan.


## The Process

<table class="cmp-table">
<tr><th>Step</th><th>What you do</th></tr>
<tr><td><strong>1. Understand the problem</strong></td><td>What's the input and output? What's genuinely difficult about it, and how might you handle that? Restating the prompt is not analysis &mdash; identifying what's hard about it is.</td></tr>

## The Key Rule: One Function, One Concern

A function isn't well decomposed just because it has a name. The real test:

<div class="pseudocode">Can you describe its job in one sentence, without joining two unrelated actions with "and"?</div>

"Reads the input *and* calculates *and* prints it" is three jobs wearing one name. These are separate concerns and belong in separate functions:

- **Reading input** &mdash; getting information from the user, a file, or an argument
- **Calculating** &mdash; pure logic, no `print()` or `input()` inside it
- **Displaying output** &mdash; formatting and showing a result, inventing nothing

Keeping these apart is what makes a function reusable: a calculation that only calculates can be tested or reused anywhere, while one tangled up with input/output is stuck serving only its original spot.

Watch both extremes:

- **Too coarse** &mdash; one function doing everything. This is the most common problem, and the one graded most strictly.
- **Too fine** &mdash; splitting one indivisible step into functions for no real reason.

Decompose along the problem's natural seams, not to hit a function count. A well-decomposed program is typically small "worker" functions (each doing exactly one of read/calculate/display) plus one coordinating function that calls them in order.


## Required Format

**1. A standard header**, as a comment block at the top of the file, with these sections in order:

- `Assignment:` the assignment name
- `Name and Loyola Id/email:`
- `Date:`
- `DESCRIPTION:` what the program does, from the user's point of view
- `QUESTIONS:` anything you're unsure about, or `None`
- `PROBLEM ANALYSIS & KEY ISSUES:` your real analysis &mdash; not a restatement of the assignment
- `PROGRAM PSEUDOCODE:` the program-level plan
- `FUNCTIONS:` one line per function, stating its single job

**2. Function skeletons**, each with a docstring describing what it does step by step &mdash; inputs, return value, essential steps. Wording can differ from the template; what matters is that every function's job is fully described before any Python logic is written inside it.

Type annotations (like `principal: float` or `-> float`) are encouraged once you're comfortable with them, but they don't replace stating what a value *means* or what's valid. Either way, the docstring should call out constraints the type alone can't express &mdash; e.g., "years must be a positive whole number." If you skip annotations, state the types themselves in the docstring instead.

## Before You Submit

- Header has all six sections, in order?
- `PROBLEM ANALYSIS & KEY ISSUES` says what's actually hard, not just the prompt restated?
- Every function's job fits one sentence, no "and" joining unrelated actions?
- Nothing written out more than once instead of being its own function?
- Written *before* the code, not described after?

See the pseudocode grading notes for how these are scored.
